```python
def number_classification():
    input1 = int(input("Enter an Integer:"))
    
    if input1 < 0:
        print ("Your number is Negative")
    elif input1 > 0: 
        print ("Your number is Positive")
    elif input1 == 0:
            print ("Your number is Zero")

    if input1 % 2 == 0:
        print("Your number is Even")
    else:
        print("Your number is Odd")
        
    if input1 % 3 == 0 and input1 % 5 == 0:
        print("Your number is divisible by both 3 and 5")
    else:
        print("Your number is not divisible by both 3 and 5")


number_classification()
```
